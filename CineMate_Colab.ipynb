```
from cinemate.data_loader import load
from cinemate.agent import CineMate
bot = CineMate(load())
print(bot.greet())
while True:
    msg = input('You: ')
    print('CineMate:', bot.reply(msg))
    if msg.strip().lower() in {'bye','quit','exit'}:
        break
```
